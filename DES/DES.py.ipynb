```python
from Crypto.Cipher import DES, AES
from Crypto.Util.Padding import pad, unpad
from Crypto.Random import get_random_bytes

# Function to perform DES encryption and decryption
def des_encrypt(plain_text, key):
    cipher = DES.new(key, DES.MODE_CBC)
    cipher_text = cipher.encrypt(pad(plain_text.encode(), DES.block_size))
    return cipher.iv + cipher_text  # Combine IV with ciphertext for decryption

def des_decrypt(cipher_text, key):
    iv = cipher_text[:DES.block_size]  # Extract IV
    cipher = DES.new(key, DES.MODE_CBC, iv)
    decrypted_data = unpad(cipher.decrypt(cipher_text[DES.block_size:]), DES.block_size)
    return decrypted_data.decode()

# Function to perform AES encryption and decryption
def aes_encrypt(plain_text, key):
    cipher = AES.new(key, AES.MODE_CBC)
    cipher_text = cipher.encrypt(pad(plain_text.encode(), AES.block_size))
    return cipher.iv + cipher_text  # Combine IV with ciphertext for decryption

def aes_decrypt(cipher_text, key):
    iv = cipher_text[:AES.block_size]  # Extract IV
    cipher = AES.new(key, AES.MODE_CBC, iv)
    decrypted_data = unpad(cipher.decrypt(cipher_text[AES.block_size:]), AES.block_size)
    return decrypted_data.decode()

# Test the functions with some data
if __name__ == "__main__":
    # DES example (Key must be 8 bytes long)
    des_key = b'21'  # 8-byte DES key
    des_plain_text = "Hello DES!"
    des_encrypted = des_encrypt(des_plain_text, des_key)
    des_decrypted = des_decrypt(des_encrypted, des_key)
    
    print("DES Encryption and Decryption:")
    print("Encrypted:", des_encrypted.hex())
    print("Decrypted:", des_decrypted)

    # AES example (Key must be 16, 24, or 32 bytes long)
    aes_key = get_random_bytes(16)  # 16-byte AES key (128-bit)
    aes_plain_text = "Hello AES!"
    aes_encrypted = aes_encrypt(aes_plain_text, aes_key)
    aes_decrypted = aes_decrypt(aes_encrypted, aes_key)
    
    print("\nAES Encryption and Decryption:")
    print("Encrypted:", aes_encrypted.hex())
    print("Decrypted:", aes_decrypted)
```
